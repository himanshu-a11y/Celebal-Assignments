# 🧠 Single Agent Pipeline Project

## Problem Statement
Build a **Single-Agent Smart Assistant** that:
- Understands user queries
- Routes tasks based on intent
- Uses tools when required
- Returns structured JSON output

### The agent should handle:
- Math queries → Calculator Tool
- Keyword extraction → Keyword Tool
- General queries → Direct response

---
### 🛠️ What You Need to Implement
- Agent logic
- Conditional routing
- Tool integration
- Basic error handling

### 🚀 Bonus
- Improve routing
- Add logging
- Add more tools


````python
# 🛠️ TOOL 1: Calculator

def calculator(expression: str) -> str:
    """Evaluate a mathematical expression."""
    try:
        # Only allow safe characters (digits, operators, parentheses, spaces, decimal points)
        allowed_chars = set("0123456789+-*/(). ")
        if not set(expression).issubset(allowed_chars):
            return "Error in calculation"
        return str(eval(expression))
    except Exception:
        return "Error in calculation"
````

````python
# 🛠️ TOOL 2: Keyword Extractor

def extract_keywords(text: str) -> list:
    """Extract keywords from text."""
    try:
        words = text.split()
        keywords = list(set([w.lower() for w in words if len(w) > 4]))
        return keywords[:5]
    except Exception:
        return []
````

## 🤖 Implement Agent Logic Below

👉 Use conditional routing:
- If query contains "calculate" → use calculator
- If query contains "keywords" → use keyword extractor
- Else → general response

````python
# 🤖 AGENT FUNCTION

import re
import logging

#  Bonus: basic logging setup
logging.basicConfig(level=logging.INFO, format="%(asctime)s | %(levelname)s | %(message)s")
logger = logging.getLogger("agent")


def agent(query: str):
    """
    Single-Agent Smart Assistant.

    Routes the incoming query to the correct tool based on intent:
      - "calculate" -> Calculator Tool
      - "keywords"  -> Keyword Extraction Tool
      - anything else -> General Response

    Returns a structured JSON-style dict:
      {"type": "calculation / keywords / general / error", "result": ...}
    """
    try:
        if not isinstance(query, str) or not query.strip():
            logger.warning("Empty or invalid query received.")
            return {"type": "error", "result": "Query cannot be empty."}

        query_lower = query.lower()
        logger.info(f"Received query: {query!r}")

        # Route 1: Calculator
        if "calculate" in query_lower:
            # Pull out the mathematical expression (numbers + operators)
            match = re.search(r"[-+*/().\d\s]+", query)
            expression = match.group().strip() if match else ""

            if not expression:
                logger.error("No valid expression found for calculation.")
                return {"type": "error", "result": "No valid expression found to calculate."}

            logger.info(f"Routing to Calculator Tool with expression: {expression!r}")
            calc_result = calculator(expression)

            if calc_result == "Error in calculation":
                return {"type": "error", "result": calc_result}
            return {"type": "calculation", "result": calc_result}

        # Route 2: Keyword Extraction
        elif "keywords" in query_lower:
            # Remove the trigger phrase
            cleaned_text = re.sub(r"extract keywords from", "", query, flags=re.IGNORECASE).strip()
            logger.info(f"Routing to Keyword Extractor Tool on text: {cleaned_text!r}")
            keywords = extract_keywords(cleaned_text)
            return {"type": "keywords", "result": keywords}

        # Route 3: General Response
        else:
            logger.info("Routing to General Response.")
            return {"type": "general", "result": f"I don't have a specialized tool for this yet, but here's your query noted: '{query}'"}

    except Exception as e:
        logger.exception("Unexpected error in agent execution.")
        return {"type": "error", "result": f"Unexpected error: {str(e)}"}
````

## 📦 Expected Output Format

```
{
  "type": "calculation / keywords / general / error",
  "result": ...
}
```

````python
# 🧪 Test Cases

queries = [
    "Calculate 20 + 5",
    "Extract keywords from Artificial Intelligence is transforming industries",
    "What is machine learning?"
]

for q in queries:
    print("Query:", q)
    print("Response:", agent(q))
    print("-" * 50)
````

````python
# 🎯 Interactive Mode

while True:
    try:
        user_input = input("Enter query (type 'exit' to stop): ")
    except Exception:

        print("No interactive input available in this environment. Exiting interactive mode.")
        break

    if user_input.lower() == "exit":
        break
    print("Response:", agent(user_input))
````
